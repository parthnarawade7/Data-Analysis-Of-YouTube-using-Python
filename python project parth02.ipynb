# **⭐Project Title: Analysis of YouTube App Usage Data using Pandas & Matplotlib::**

# 📌 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

🔍 Explanation

pandas → used for creating tables (DataFrames), cleaning data, and analysis.

matplotlib.pyplot → used for creating graphs such as line charts, bar charts, scatter plots.

These libraries are the basic tools for any data analysis project.**bold text**

# 📌 Cell 2 — Creating the Dataset

```python
data = {
    'Day': ['Mon','Tue','Wed','Thu','Fri','Sat','Sun'],
    'Watch_Time_Min': [120, 95, 110, 150, 180, 240, 300],
    'Videos_Watched': [15, 10, 13, 17, 22, 30, 35],
    'Data_Used_MB': [500, 420, 460, 600, 750, 1000, 1250]
}

df = pd.DataFrame(data)
df
```

🔍 Explanation

We create a dictionary named data that stores:

Days of the week

Watch time (in minutes)

Number of videos watched

Data usage (in MB)

Then we convert the dictionary into a DataFrame using pd.DataFrame().

Displaying df shows our dataset in table form.

This is basically our "input data" for analysis.**bold text**

# 📌 Cell 3 — Basic Analysis of the Dataset

```python
print("Basic Info:")
print(df.info())

print("\nSummary Statistics:")
print(df.describe())
```

🔍 Explanation

df.info() → tells structure of the dataset:

Number of rows

Number of columns

Data types (int, float, string, etc.)

df.describe() → gives statistics:

Mean

Minimum

Maximum

Standard deviation

Quartiles

This helps you understand the overall pattern of the data.**bold text**

# 📌 Cell 4 — Line Chart (Watch Time Over the Week)

```python
plt.figure(figsize=(8,5))
plt.plot(df['Day'], df['Watch_Time_Min'])
plt.xlabel("Day")
plt.ylabel("Watch Time (Minutes)")
plt.title("YouTube Watch Time Per Day")
plt.show()
```

🔍 Explanation

plt.figure() → sets the size of the graph.

plt.plot() → creates a line graph.

X-axis → Days

Y-axis → Watch time in minutes

plt.xlabel() and plt.ylabel() → label the axes.

plt.title() → gives a title to the graph.

plt.show() → displays the graph.

This graph shows how the user’s watch time changes throughout the week.**bold text**

# 📌 Cell 5 — Bar Chart (Videos Watched)

```python
plt.figure(figsize=(8,5))
plt.bar(df['Day'], df['Videos_Watched'])
plt.xlabel("Day")
plt.ylabel("Videos Watched")
plt.title("Videos Watched Per Day")
plt.show()
```

🔍 Explanation

plt.bar() creates a bar graph.

It visually compares the number of videos watched on each day.

Bar charts are good for comparing categories (days in this case).**bold text**

# 📌 Cell 6 — Scatter Plot (Watch Time vs Data Used)

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Watch_Time_Min'], df['Data_Used_MB'])
plt.xlabel("Watch Time (Minutes)")
plt.ylabel("Data Used (MB)")
plt.title("Correlation Between Watch Time & Data Usage")
plt.show()
```

🔍 Explanation

plt.scatter() → creates a scatter plot.

Shows relationship between:

Watch time

Data used

Helps us understand:

Higher watch time = higher data consumption

Whether two variables are correlated

Scatter plots are best to show such relationships.**bold text**

# 📌 Cell 7 — Finding the Day With Maximum Usage

```python
max_day = df.loc[df['Watch_Time_Min'].idxmax()]
print("Day with Maximum Watch Time:")
print(max_day)
```

🔍 Explanation

df['Watch_Time_Min'].idxmax()
→ finds the index of the highest watch time value.

df.loc[...]
→ retrieves the full row of that day.

This tells which day the app usage is highest.

For example, Sunday has the highest watch time.**bold text**
